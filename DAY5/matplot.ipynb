```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# hand made data set
price = [20000,25000,18000,15000,800000]
year = [2000,2001,2002,2003,2004]
plt.plot(year,price)
plt.xlabel("price")
plt.ylabel("Year")
plt.title("Year Vs Price")
plt.show()
```

```python
# importing data
data = pd.read_csv('sharma-kohli.csv')
```

```python
data.head()
```

```python
# plotting virat kohli data
plt.plot(data['index'],data['V Kohli'],marker='d')
plt.xlabel("Year")
plt.ylabel("Runs")

plt.title("Year wise Kohli")
```

```python
#plotting RG Sharma data
plt.plot(data['index'],data['RG Sharma'],marker='o')
plt.xlabel("Year")
plt.ylabel("Runs")

plt.title("Year wise Sharma")
```

```python
# plotting combined data of Virat and Sharma
plt.plot(data['index'],data['RG Sharma'],marker='o')
plt.plot(data['index'],data['V Kohli'],marker='o')
plt.xlabel("Year")
plt.ylabel("Runs")

plt.title("Year wise Kohli and Sharma")
```

```python
from matplotlib.lines import lineStyles
# plotting combined data of Virat and Sharma with extra properties
plt.plot(data['index'],data['RG Sharma'],marker='o',color='#D9F10F',linestyle='solid',linewidth=3,markersize=8,label='Virat')
plt.plot(data['index'],data['V Kohli'],marker='o',color='#FC00D6',linestyle='dashed',linewidth=3,markersize=8,label='Rohit')
plt.xlabel("Year")
plt.ylabel("Runs")
plt.legend() # we can direction also as upper right, upper left etc
plt.title("Year wise Kohli and Sharma")
```

```python

# plotting combined data of Virat and Sharma with Grid
plt.plot(data['index'],data['RG Sharma'],marker='o',color='#D9F10F',linestyle='solid',linewidth=3,markersize=8,label='Virat')
plt.plot(data['index'],data['V Kohli'],marker='o',color='#FC00D6',linestyle='dashed',linewidth=3,markersize=8,label='Rohit')
plt.xlabel("Year")
plt.ylabel("Runs")
plt.legend() # we can direction also as upper right, upper left etc
plt.title("Year wise Kohli and Sharma")
plt.grid()
```

```python
df=pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%')

plt.show()
```

```python
# percentage and colors
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['blue','green','yellow','pink','cyan','brown'])
plt.show()
```

```python
# explode
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.1,0,0,0,0,1],shadow=True)
plt.show()
```
