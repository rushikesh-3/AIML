### Pandas


```python
# pandas are used to storing tabular data.
# Pandas revolves around only TWO core structures: Series, DataFrames
# Series     = One Column
# DataFrame  = Collection of Series
# A DataFrame is a collection of Series arranged as rows and columns.
# Pandas helps store, organize, filter, sort, analyze, and manipulate large tabular datasets efficiently.
```

```python
# Method 1: Dictionary

import pandas as pd

students={
    "name":["Rushi","Raju","Rakesh"],
    "Age":[20,21,22],
    "Marks":[80,85,90]
}

df=pd.DataFrame(students)
```

```python
print(df)
```

```python
# Method 2: List of Dictionaries

data=[
    {"name":"Rushi","age":20},
    {"name":"Raju","age":21},
    {"name":"Rani","age":22},
]

df1=pd.DataFrame(data)
```

```python
print(df1)
```

```python
# Method 3: List of Lists

data1=[
    ["Rahul",20],
    ["Ram",22]
]

df2=pd.DataFrame(
    data1,
    columns=["Name","Age"]
)
```

```python
print(df2)
```
